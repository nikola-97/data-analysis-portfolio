```python
# Import necessary libraries and read the CSV file [19.11.2026]
import pandas as pd

df = pd.read_csv(r"C:\Users\yalci\Documents\Raw Data Files\Uncleaned Data\Iranian_Transactions.csv")
df.head()
```

```python
# Get some information about the DataFrame
df.info()
```

```python
# Unique values in the 'status' column
df['status'].unique()
```

```python
# Standardize the 'status' column
df['status'] = df['status'].replace({
                                        'succeed': 'success', 'Success': 'success',
                                        'failed': 'fail', 'FAIL': 'fail'})
df['status'].unique()
```

```python
# Unique values in the 'card_type' column
df['card_type'].unique()
```

```python
# Standardize the 'card_type' column
df['card_type'] = df['card_type'].replace({
                                            'visa': 'Visa', 'VISA': 'Visa', 'Vsa': 'Visa',
                                            'MastCard': 'Master Card', 'MasterCard': 'Master Card',
                                            'Master-Card': 'Master Card'})
df['card_type'].unique()
```

```python
# Unique values in the 'city' column
df['city'].unique()
```

```python
# Standardize the 'city' column
df['city'] = df['city'].replace({
                                    'Tehran ': 'Tehran', 'TEHRAN': 'Tehran',
                                    'THR': 'Tehran', 'tehr@n': 'Tehran', 
                                    'ThRan': 'Tehran','karaj': 'Karaj'})
df['city'].unique()
```

```python
# Check if transactions have unique id
len(df) == df["id"].nunique()

print("Rows:", len(df))
print("Unique IDs:", df["id"].nunique())
```

```python
# Many transactions have the same id
df[df["id"] == 1]
```

```python
df["id"].value_counts()[df["id"].value_counts() > 1]
```

```python
# Total number of duplicate transactions
print("Rows:", len(df))
print("Duplicate transactions:", df.duplicated(
    subset=[
        "status",
        "time",
        "card_type",
        "city",
        "amount"
    ]
).sum())    # which implies that there are 9 974 unique transactions
```

```python
# Drop duplicates and 'id' column as unreliable identifier
df = df.drop_duplicates(
    subset=[
        "status",
        "time",
        "card_type",
        "city",
        "amount"
    ]
).drop(columns=["id"])

df
```

```python
# Split the column 'time'
df[["trans_date", "trans_time"]] = df["time"].str.split(" ", expand=True)
df
```

```python
# Drop the column 'time' as it is no longer needed
df = df.drop(columns=["time"])
df
```

```python
# Order the df by 'trans_date' and 'trans_time' and then add a new column 'trans_id'
df = df.sort_values(["trans_date", "trans_time"])
df.insert(0, "trans_id", range(1, len(df) + 1))
df
```

```python
# Reset the index of the DataFrame
df.reset_index(drop=True, inplace=True)
df
```

```python
# Check if trans_id is unique
df["trans_id"].is_unique
```

```python
# Set trans_id as the index of the DataFrame
df["trans_id"] = df.index + 1
df.set_index("trans_id", inplace=True)
df
```

```python
# Format the 'trans_date' column to datetime format
df["trans_date"] = pd.to_datetime(df["trans_date"], format="%Y-%m-%d")

# Format the 'trans_time' column to time format
df["trans_time"] = pd.to_datetime(df["trans_time"], format="%H:%M:%S").dt.time

df
```

```python
# Format the 'amount' column with two decimal places
df_clean = df.copy()
df_clean["amount"] = pd.to_numeric(df_clean["amount"], errors="coerce").round(2)
df_clean
```
